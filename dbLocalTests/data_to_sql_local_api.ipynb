```python
payload = {
        "domain": [("active", "!=", False)],
        "fields": [
            'id',
            'name',
            'job_title',
            'department_id',
            'x_hire_date',
            'x_warehouse_id',
        ],
        "context": {"lang": "es_MX"},
    }

empleados_data = api.search_read('hr.employee', payload=payload)
```

```python
empleados_df = pd.DataFrame(empleados_data)

empleados_df.rename(columns={
    'id': 'id_odoo',
    'x_hire_date': 'hire_date_str',
}, inplace=True)

empleados_df['department'] = empleados_df['department_id'].apply(lambda x: 'Ventas Piso' if 'Ventas Piso' in x[1] else x[1])
empleados_df['warehouse'] = empleados_df['x_warehouse_id'].apply(lambda x: 'A1' if x[0]==2 else 'A2')
empleados_df.drop(columns=['department_id', 'x_warehouse_id'], inplace=True)
```

```python
endpoint = url + '/empleados/new'

ingresados = []
id_errores = []

for i, row in empleados_df.iterrows():
    empleado = row.fillna(None).to_dict()
    x = requests.post(endpoint, json = empleado)
    ingresados.append(x.ok)
    if x.ok:
            ingresados.append(x.ok)
    else:
        print(x, ' para ', row['id_odoo'])
        print(x.json())
        id_errores.append(row.name)

print(f'{len(ingresados)} de {len(empleados_df)} empleados ingresados con éxito')
```

```python
payload = {
        "domain": [("type", "=", "consu"), ("company_id", "=", False)],
        "fields": [
            'id',
            'name',
            'default_code',
            'lst_price',
            'standard_price',

        ],
        "context": {"lang": "es_MX"},
    }

productos_data = api.search_read('product.product', payload=payload)
```

```python
productos_df = pd.DataFrame(productos_data)

productos_df.rename(columns={
    'id': 'id_odoo',
    'name': 'name',
    'default_code': 'code',
    'lst_price': 'price',
    'standard_price': 'cost',
}, inplace=True)
```

```python
endpoint = url + '/productos/new'

ingresados = []
id_errores = []

for i, row in productos_df.iterrows():
    producto = row.fillna(None).to_dict()
    x = requests.post(endpoint, json = producto)
    ingresados.append(x.ok)
    if x.ok:
        ingresados.append(x.ok)
    else:
        print(x, ' para ', row['id_odoo'])
        print(x.json())
        id_errores.append(row.name)

print(f'{len(ingresados)} de {len(productos_df)} productos ingresados con éxito')
```
